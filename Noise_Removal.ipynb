```python
import cv2
import matplotlib.pyplot as plt

# Load image
image_path = "Skin Disease Dataset/Dermatofibroma/ISIC_0011410.jpg"   # give your image path
img = cv2.imread(image_path)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# -----------------------------
# Apply noise removal methods
# -----------------------------

# 1. Gaussian Blur (smooth noise)
gaussian = cv2.GaussianBlur(img, (5,5), 0)

# 2. Median Filter (salt-pepper noise)
median = cv2.medianBlur(img, 5)

# 3. Bilateral Filter (preserves edges)
bilateral = cv2.bilateralFilter(img, 9, 75, 75)

# -----------------------------
# Display results
# -----------------------------
plt.figure(figsize=(12,8))

plt.subplot(2,2,1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(2,2,2)
plt.imshow(gaussian)
plt.title("After Gaussian Noise Removal")
plt.axis("off")

plt.subplot(2,2,3)
plt.imshow(median)
plt.title("After Median Noise Removal")
plt.axis("off")

plt.subplot(2,2,4)
plt.imshow(bilateral)
plt.title("After Bilateral Noise Removal")
plt.axis("off")

plt.tight_layout()
plt.show()
```
